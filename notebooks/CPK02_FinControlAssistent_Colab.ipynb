```python
from pathlib import Path

arquivos = [
    Path("app/vector_store.py"),
    Path("app/rag.py"),
    Path("notebooks/CPK02_DocMind_RAG.ipynb"),
]

termos = [
    "OllamaEmbeddings",
    "ChatOllama",
    "nomic-embed-text",
    "gemma4:cloud",
    "base_url",
    "OLLAMA_API_KEY",
    "ollama pull",
    "localhost",
]

for arquivo in arquivos:
    print(f"\n{'=' * 20} {arquivo} {'=' * 20}")

    if not arquivo.exists():
        print("Arquivo não encontrado.")
        continue

    if arquivo.suffix == ".ipynb":
        import json
        notebook = json.loads(arquivo.read_text(encoding="utf-8"))
        linhas = []

        for i, celula in enumerate(notebook.get("cells", [])):
            if celula.get("cell_type") == "code":
                codigo = "".join(celula.get("source", []))
                if any(t.lower() in codigo.lower() for t in termos):
                    linhas.append(f"[Célula {i}]\n{codigo}")

        print("\n\n".join(linhas) if linhas else "Nenhum trecho relevante encontrado.")
    else:
        linhas = arquivo.read_text(encoding="utf-8").splitlines()

        indices = [
            i for i, linha in enumerate(linhas)
            if any(t.lower() in linha.lower() for t in termos)
        ]

        exibidos = set()
        for i in indices:
            for j in range(max(0, i - 2), min(len(linhas), i + 3)):
                if j not in exibidos:
                    print(f"{j + 1}: {linhas[j]}")
                    exibidos.add(j)
            print("---")
```

```python
from app.vector_store import buscar, nome_do_arquivo

pergunta = "Quanto devo ter de reserva de emergência?"

for tamanho in (512, 1024):
    print(f"\n{'=' * 15} CHUNK {tamanho} {'=' * 15}")

    resultados = buscar(
        pergunta,
        tamanho_chunk=tamanho,
        k=3
    )

    for i, (documento, similaridade) in enumerate(resultados, start=1):
        print(f"\nResultado {i}")
        print("Fonte:", nome_do_arquivo(documento))
        print("Página:", documento.metadata.get("page", "N/A"))
        print(f"Similaridade: {similaridade:.4f}")
        print("Trecho:", documento.page_content[:700])
```
